## 2. Fields: feature / label / context / excluded

### Features

The features I plan to use are:

- `impressions_90d` — recent search exposure available before the refresh decision.
- `clicks_90d` — recent click performance available before the decision.
- `ctr` — observed click-through performance available before the decision.
- `avg_position` — search ranking information available before the decision.
- `days_since_last_update` — freshness information available before the decision.

### Label / proxy

- **Refresh opportunity proxy** — a proxy representing the potential value of refreshing a page. The ideal label would be the improvement in performance after a refresh, but the current warehouse data does not directly provide that post-refresh outcome.

### Context

These fields can provide useful context about the page or help analyze segments, but I will not use them as model features in the five-feature frame:

- `content_id` — identifies the content item.
- `client_id` — identifies the client/account.
- `content_type` — describes the type of content.
- `main_intent` — describes the search intent.
- `competition_level` — describes the competitive environment.
- `age_tier` — groups content by age.
- `freshness_tier` — groups content by freshness.
- `trend_direction` — describes the current performance trend.

### Excluded

- `impressions_last_30d`, `clicks_last_30d`, and `sessions_last_30d` from a future/outcome period will be excluded when they are not available at the decision moment.
- Any future-month performance fields will be excluded because they contain information that would only be known after the refresh decision and could cause label leakage.
- Any field directly constructed from the future outcome/label will also be excluded for the same reason.

## 4. Data limits

This data cannot tell me with certainty whether refreshing a page will actually cause its performance to improve. The historical data is observational, so it does not provide a controlled before/after experiment for every content refresh.

There are also several limitations:

- **Unbalanced history:** Some content has more complete performance history than other content, so comparisons between pages may not be equally reliable.
- **GSC-only early rows:** Some earlier records may have GSC data available without equivalent GA4 data, which limits the completeness of engagement-based features.
- **Window overlaps:** The 90-day windows overlap across periods, so observations are not fully independent. A later window can contain much of the same historical information as an earlier window.
- **Missing causal information:** The data does not fully capture why a page performed poorly or whether a specific content change caused a later improvement.

Therefore, the model can rank pages by estimated refresh opportunity, but it cannot guarantee that a refresh will improve performance.
