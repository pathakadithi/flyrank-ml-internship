### Task type: Scoring

My lane is Refresh / Content Opportunity Scoring.

The goal is to assign each content page a score representing how strong the opportunity is for a content refresh. The pages can then be prioritized from highest to lowest refresh opportunity.

This is a scoring problem because the goal is not only to predict whether a page should be refreshed or not. Instead, we want to produce a continuous priority score that helps rank many pages and identify which pages should be considered for refresh first.

### Target / Proxy

The ideal target would be the improvement in a page's performance after it is refreshed, such as the change in CTR or clicks after the refresh. However, the starter dataset does not contain an observed post-refresh outcome for each page.

Therefore, for this initial framing, I would treat the target as a proxy for refresh opportunity rather than a true supervised target. The proxy can be constructed from signals such as current CTR, impressions, clicks, search volume, competition, and content characteristics.

The goal of the proxy is to estimate which pages have the strongest potential opportunity for improvement. A future production model should replace this proxy with an observed outcome from historical before-and-after content refreshes.

## 3) Success metric

The primary success metric for this scoring task would be **Precision@K**, where K represents the number of pages selected for refresh.

For example, Precision@20 would measure the proportion of the top 20 pages ranked by the model that actually benefit from a content refresh.

This metric fits the problem because the content team has limited time and cannot refresh every page. The model should therefore place the pages with the strongest refresh opportunities near the top of the ranking.

In a production setting, the final outcome could be measured using the actual improvement in CTR, clicks, or another agreed-upon performance measure after the selected pages are refreshed.

The key success criterion is therefore:

**The higher the precision among the top-ranked pages, the more useful the model is for prioritizing content refresh work.**

### Success Metric: Precision@K

I would use Precision@K as the main success metric because the model's purpose is to prioritize pages for a limited content refresh queue.

Precision@K measures how many of the pages selected among the top K scored pages are actually valuable refresh opportunities.

For example, if the content team selects the top 100 pages from the model's ranking, a good model should have a high proportion of genuinely valuable refresh opportunities within those 100 pages.

This metric is useful because it directly measures whether the highest-priority pages selected by the scoring system are useful for the real content action.

## 5) Why ML beats a fixed rule here

A fixed rule could identify refresh candidates using a simple threshold, such as refreshing every page with a CTR below a certain value. However, this would ignore the interaction between multiple signals.

For example, a page may have a low CTR but also have very few impressions, making it a lower-priority refresh candidate. Another page may have high impressions, declining performance, an older content age, and a long time since its last update, making it a stronger opportunity.

The dataset contains multiple signals including impressions, clicks, CTR, average position, content age, days since last update, freshness, and performance trends. An ML-based scoring approach can combine these signals and learn patterns associated with stronger refresh opportunities instead of relying on manually selected thresholds.

This is especially useful when the number of pages is large and the content team has limited capacity. The model can rank pages by opportunity so that the highest-priority pages can be reviewed and refreshed first.

ML is therefore useful because the goal is not simply to apply one fixed rule, but to learn a more flexible ranking of content opportunities from multiple signals.
